# Hypothese 2

Die Ergebnisse zeigen, dass Regularisierung die Generalisierung verbessert und Overfitting reduziert.

- Baseline: Train = 0.9955, Test = 0.9864, Gap = 0.0091
- Dropout: Train = 0.9770, Test = 0.9884, Gap = -0.0114
- L2-Regularisierung: Train = 0.9833, Test = 0.9822, Gap = 0.0010
- Data Augmentation: Train = 0.9786, Test = 0.9839, Gap = -0.0053


Fazit: Das Dropout-Modell hat in diesem Lauf die beste Testgenauigkeit erreicht, waehrend L2-Regularisierung den kleinsten positiven Train-Test-Abstand zeigt.

##  Visualisierungen

### 1. accuracy_comparison.png
Dieser Plot zeigt die Trainingsgenauigkeit und Testgenauigkeit der 4 Modelle.

- **Was zeigt der Plot?**
  Er zeigt, wie gut jedes Modell auf bekannten Daten (Train) und auf unbekannten Daten (Test) abschneidet.
- **Warum ist er wichtig?**
  Er ist der Hauptvergleich zwischen den Varianten (Baseline, Dropout, L2, Augmentation).
- **Welche Information nehmen wir mit?**
  Wir sehen direkt, welches Modell die beste Testleistung erreicht und ob ein Modell nur auf den Trainingsdaten stark ist.

### 2. before_after_test_accuracy.png
Dieser Plot vergleicht die Testgenauigkeit vor dem Training mit der Testgenauigkeit nach dem Training.

- **Was zeigt der Plot?**
  Er zeigt den Lernfortschritt jedes Modells auf denselben Testdaten.
- **Warum ist er wichtig?**
  Er macht sichtbar, dass das Training tatsächlich einen deutlichen Leistungsgewinn bringt.
- **Welche Information nehmen wir mit?**
  Der Abstand zwischen "Before" und "After" zeigt, wie stark das Modell durch Training verbessert wurde.

### 3. overfitting.png
Dieser Plot zeigt pro Modell den Overfitting-Indikator zwischen den Train und Test 

- **Was zeigt der Plot?**
  Positive Werte bedeuten: Training besser als Test.
- **Warum ist er wichtig?**
  Er quantifiziert Overfitting in einer einzelnen Zahl pro Modell.
- **Welche Information nehmen wir mit?**
  Je kleiner der positive Abstand, desto besser ist die Generalisierung. Sehr große positive Werte sind ein Warnzeichen.

### 4. generalization_gap.png
Dieser Plot zeigt ebenfalls die Generalization Gap zwischen den Train und Test, mit Fokus auf die Stabilität der Modelle auf unbekannten Daten.

- **Was zeigt der Plot?**
  Er visualisiert den Unterschied zwischen Lernleistung auf Trainingsdaten und Leistung auf Testdaten.
- **Warum ist er wichtig?**
  Er zeigt, ob ein Modell robust verallgemeinert oder zu stark auf Trainingsdaten spezialisiert ist.
- **Welche Information nehmen wir mit?**
  Modelle mit Gap nahe 0 generalisieren in der Regel besser; größere Gaps deuten auf Overfitting-Tendenz hin.
